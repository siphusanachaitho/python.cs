```python

fruits = []
adj = []
j=0
while True:
    inp_state = input("press q for end ,or Enter to Any Input")
    if(inp_state.lower() == 'q'):
        break

    else:
      inp1 = input("Fruit name is :")
      fruits.append(inp1)

      inp2 = input("Color name is :")
      adj.append(inp2)

```

```python
#Lamda 
x = lambda a : a *10
print(x(5))

x = lambda x1, y1, x2,y2, x3,y3 :  0.5 * abs(x1*(y2-y3)+ x2*(y3-y1)+ x3*(y1-y2))

print(x(1,1, 4,1, 1,5))
```
